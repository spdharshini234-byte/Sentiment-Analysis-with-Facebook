Build TF-IDF features on the training split only and inspect the vocabulary.

```python
import sys, json
from pathlib import Path
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style="whitegrid")
from src import config
from src.data.load_data import load_raw, inspect_dataset
from src.data.preprocess import prepare_dataframe, clean_text
from src.models.train import grouped_stratified_split
from src.features.tfidf_features import build_vectorizer
```

```python
df, _ = prepare_dataframe(load_raw())
tr, va, te = grouped_stratified_split(df)
vec = build_vectorizer()
X = vec.fit_transform(df.iloc[tr]['clean_text'])
print('train matrix:', X.shape, '| non-zero:', X.nnz)
```

```python
names = vec.get_feature_names_out()
scores = np.asarray(X.mean(axis=0)).ravel()
pd.DataFrame({'feature': names, 'mean_tfidf': scores}).sort_values('mean_tfidf', ascending=False).head(15)
```
